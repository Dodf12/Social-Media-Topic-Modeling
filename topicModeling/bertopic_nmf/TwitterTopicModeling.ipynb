```python
import re

def clean_tweet(tweet, keywords):
  tweet = str(tweet).lower()

  #remove dominant words
  for k in keywords:
    tweet = tweet.replace(k, "")

  #remove months
  # use re to remove just the month
  tweet = re.sub(r"\bjan\b", '', tweet)
  tweet = re.sub(r"\bfeb\b", '', tweet)
  tweet = re.sub(r"\bmar\b", '', tweet)
  tweet = re.sub(r"\bapr\b", '', tweet)
  tweet = re.sub(r"\bmay\b", '', tweet)
  tweet = re.sub(r"\bjun\b", '', tweet)
  tweet = re.sub(r"\bjul\b", '', tweet)
  tweet = re.sub(r"\baug\b", '', tweet)
  tweet = re.sub(r"\bsep\b", '', tweet)
  tweet = re.sub(r"\bnov\b", '', tweet)
  tweet = re.sub(r"\bdec\b", '', tweet)

  #remove @ mention
  x = 0
  val = tweet
  while val.find('@', x+1) != -1:
    x = val.find('@', x+1)

  if x != -1:
    x = val.find(' ', x+1)
    if x!= -1:
      tweet = val[x:]

  #remove retweeted, spaces, urls, and other mentions
  tweet = re.sub(r"http\S+|www\.\S+|\b\w+\.(?:ly|com|org|net|io)\S*", " ", tweet)
  tweet = re.sub(r"\bretweeted\b", " ", tweet, flags=re.IGNORECASE)
  tweet = re.sub(r"@\w+", " ", tweet)
  tweet = re.sub(r"\s+", " ", tweet).strip()
  return tweet


def clean_data(file, keywords):
  # only want text since BERTopic looks
  file = file.drop(columns="ID")
  docs = []
  for string in file["Text"]:
    docs.append(clean_tweet(string, keywords))


  return docs
```

```python
import numpy as np
import nltk

from bertopic import BERTopic
from nltk.tokenize import word_tokenize
from nltk import pos_tag
```

```python
def nouns_adj(text):
    is_noun_adj = lambda pos: pos[:2] == 'NN' or pos[:2] == 'JJ'
    tokenized = word_tokenize(text)
    return ' '.join([word for word, pos in pos_tag(tokenized) if is_noun_adj(pos)])

def clean_data_nmf(file, keywords):
  # cleans data for NMF
  file = file.drop(columns="ID")
  docs = []
  for string in file["Text"]:
    string = clean_tweet(string, keywords)
    string = nouns_adj(string)
    docs.append(string)

  return docs
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import NMF

nltk.download("punkt")
nltk.download('averaged_perceptron_tagger_eng')
```

```python
def run_nmf(file, keys, numTop):
  docs = clean_data_nmf(file, keys)

  # vectorize (convert to vectors)
  vectorizer = TfidfVectorizer(
      stop_words="english",
      max_df = 0.8,
      min_df = 2
  )
  X = vectorizer.fit_transform(docs)

  nmf = NMF(n_components=numTop, random_state = 42)
  W = nmf.fit_transform(X)
  H = nmf.components_

  topic_words = []

  vocab = vectorizer.get_feature_names_out()
  for k in range(numTop):
    top_idx = H[k].argsort()[::-1][:10]
    words = [vocab[i] for i in top_idx]
    topic_words.append(words)
  doc_topics = W.argmax(axis=1)
  return doc_topics, topic_words

```

```python
import re

def clean_for_bertopic(df, text_column="Text", min_words=10):
    df = df.copy()
    text = df[text_column].fillna("")

    # Fix escaped / literal newlines
    text = text.str.replace(r"\\n", " ", regex=True)
    text = text.str.replace(r"\n", " ", regex=True)

    # Remove URLs
    text = text.str.replace(
        r"https?://\S+|www\.\S+|https?\s*//\S+|\S+\.com\S*|\S+\.org\S*|\S+\.io\S*|\S+\.co\S*/\S*",
        "", regex=True,
    )

    # Remove email addresses
    text = text.str.replace(r"\S+@\S+\.\S+", "", regex=True)

    # Decode common unicode escape sequences
    def _decode_unicode(s):
        try:
            return s.encode("utf-8").decode("unicode_escape", errors="ignore")
        except (UnicodeDecodeError, UnicodeEncodeError):
            return re.sub(r"\\u[0-9a-fA-F]{4}", "", s)

    text = text.apply(lambda x: _decode_unicode(x) if isinstance(x, str) else x)

    # Remove non-ASCII symbols & emoji
    text = text.str.replace(r"[^\x20-\x7E]", " ", regex=True)

    # Remove formatting artifacts
    text = text.str.replace(r"[_]{3,}", " ", regex=True)
    text = text.str.replace(r"[-]{3,}", " ", regex=True)
    text = text.str.replace(r"\*{1,3}", " ", regex=True)
    text = text.str.replace(r"\\+", " ", regex=True)

    # Remove Reddit user/subreddit mentions
    text = text.str.replace(r"/?[ur]/\w+", "", regex=True)

    # Remove isolated special characters
    text = text.str.replace(r"[|^{}\[\]<>~`@=#]", " ", regex=True)

    # Collapse whitespace
    text = text.str.replace(r"\s+", " ", regex=True)
    text = text.str.strip()

    df[text_column] = text

    # Drop exact duplicates
    df = df.drop_duplicates(subset=text_column).reset_index(drop=True)

    # Drop short rows (bumped from 5 to 10)
    df = df[df[text_column].str.split().str.len() >= min_words].reset_index(drop=True)

    return df
```

```python
import re
from sklearn.cluster import KMeans

def clean_negative_tweet(tweet, keywords):
  tweet = str(tweet).lower()

  #remove dominant words
  for k in keywords:
    #tweet = tweet.replace(k, "")
    tweet = re.sub(r'\b' + k + r'\b', '', tweet)


  #remove months
  # use re to remove just the month
  tweet = re.sub(r"\b(jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\b", "", tweet)
  tweet = re.sub(r"\b\d+\b", "", tweet)

  #remove @ mention
  x = 0
  val = tweet
  while val.find('@', x+1) != -1:
    x = val.find('@', x+1)

  if x != -1:
    x = val.find(' ', x+1)
    if x!= -1:
      tweet = val[x:]

  #remove retweeted, spaces, urls, and other mentions
  tweet = re.sub(r"http\S+|www\.\S+|\b\w+\.(?:ly|com|org|net|io)\S*", " ", tweet)
  tweet = re.sub(r"\bretweeted\b", " ", tweet, flags=re.IGNORECASE)
  tweet = re.sub(r"@\w+", " ", tweet)
  tweet = re.sub(r"\s+", " ", tweet).strip()
  tweet = re.sub(r"\b(newest|load)\b", "", tweet)

  return tweet


def clean_negative_data(file, keywords):
  # only want text since BERTopic looks
  # file = file.drop(columns="tweet_name")
  docs = []
  for string in file["Text"]:
    docs.append(clean_negative_tweet(string, keywords))

  docs = list(set(docs))

  docs = [d for d in docs if len(d.split()) >= 8]
  docs = [d for d in docs if not re.search(r'\b(que|la|le|je|est|les|não|eu|aí)\b', d)]
  docs = [d for d in docs if not re.search(r'(make money|piped\.video|retweet.*tag|stock|\$[A-Z]{2,}|petition|signatures|newest|load|enable hls playback)', d, re.IGNORECASE)]
  docs = [d for d in docs if not re.search(
      r'(make money|piped\.video|retweet.*tag|\$[A-Z]{2,}|petition|signatures|'
      r'newest|load|enable hls playback|账号|批发|电报|earncrypto|cleanapp|'
      r'tokenization|blockchn|crispr|cloning|dna|genome|ballabon|medvin)',
      d, re.IGNORECASE
  )]

  return docs

def run_negative(file, keywords, n_topics = 8):

  custom_stops = list(CountVectorizer(stop_words="english").get_stop_words())
  custom_stops += ['like', 'just', 'don', 'use', 'know', 'really',
                     'want', 'think', 've', 'good', 'going', 'new',
                     'things', 'better', 'need', 'way', 'using', 'help',
                     'does', 'right', 'lot', 'time', 'say', 'feel',
                     'make', 'actually', 'tell', 'chat',
                     'ai', 'chatgpt', 'gpt', 'therapist', 'therapists',
                     'therapy', 'mental', 'health', 'human', 'humans',
                     'people', 'person', 'bot', 'chatbot']

  vectorizer_model = CountVectorizer(stop_words=custom_stops, min_df=2)
  cluster_model = KMeans(n_clusters=n_topics, random_state=42)

  topic_model = BERTopic(embedding_model="all-MiniLM-L6-v2",
                         vectorizer_model=vectorizer_model,
                         hdbscan_model=cluster_model,
                         )
  docs = clean_negative_data(file, keywords)

  # embed
  topics, prob = topic_model.fit_transform(docs)
  topics = topic_model.topics_

  return topics, prob, topic_model.get_topic_info(), topic_model, docs
```

```python
def negative_nouns_adj(text):
    is_noun_adj = lambda pos: pos[:2] == 'NN' or pos[:2] == 'JJ'
    tokenized = word_tokenize(text)
    return ' '.join([word for word, pos in pos_tag(tokenized) if is_noun_adj(pos)])

def neg_clean_data_nmf(file, keywords):
  # cleans data for NMF
  #file = file.drop(columns="ID")
  docs = []
  for string in file["Text"]:
      text = str(string).lower()

      for k in keywords:
          #text = text.replace(k, "")
          text = re.sub(r'\b' + k + r'\b', '', text)


      text = re.sub(r"^.*?\d{4}\s*", "", text)  # remove username/date prefix

      text = re.sub(r"http\S+|www\.\S+|\b\w+\.(?:ly|com|org|net|io)\S*", " ", text)
      text = re.sub(r"[^\x20-\x7E]", " ", text)
      text = re.sub(r"\b\d+\b", " ", text)
      text = re.sub(r"[|^{}\[\]<>~`@=#]", " ", text)
      text = re.sub(r"\s+", " ", text).strip()

      text = re.sub(r"\b(jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\b", "", text)
      text = re.sub(r"\b(enable|hls|playback|dlvr|retweeted|newest|load)\b", "", text)
      text = re.sub(r"\b(amasad|amjad|masad|joshua|mnstream|mainstream|solopreneur|aisolopreneur|lilianweng|deaner444|sd|bitch|reporter|basic|grok|danmurrayserter|robertrmorris|christinelu|taiwanese|deanehenryson|nickadobos|nickdobos|dobos|nick|lilian|weng)\b", "", text)
      text = re.sub(r"\b\w{1,2}\b", "", text)  # remove 1-2 character words
      text = re.sub(r"\b\w*_\w*\b", "", text)  # remove words with underscores
      text = re.sub(r"\b(danmurrayserter|robertrmorris)\b", "", text)
      text = re.sub(r"\s+", " ", text).strip()

      string = negative_nouns_adj(text)
      docs.append(string)

  docs = list(set(docs))
  docs = [d for d in docs if len(d.split()) >= 5]

  docs = [d for d in docs if not re.search(
    r'(memecoin|airdrop|piped|retweet|earn.*art|beana|creator.*army|bodycalc|transitionai|visitor.*creator|goodbye.*jobs|money.*steps|'
    r'aitherapy.*army|crypto|sexyai|aisexy|aiart|aiholidayjoy|merrychristmas|peaceandlove|christmas.*joy|platform.*creator|micky|'
    r'stock.*price|sgn|vygr|doubled|trading|beginner.*llm.*mom|notion.*site.*beginn)',
    d, re.IGNORECASE
    )]

  return docs
```

```python
from nltk.tokenize import word_tokenize
from nltk import pos_tag
import nltk
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import NMF

def run_nmf(file, keys, numTop):
  docs = neg_clean_data_nmf(file, keys)

  # vectorize (convert to vectors)
  vectorizer = TfidfVectorizer(
      stop_words="english",
      max_df = 0.8,
      min_df = 2
  )
  X = vectorizer.fit_transform(docs)

  nmf = NMF(n_components=numTop, random_state = 42)
  W = nmf.fit_transform(X)
  H = nmf.components_

  topic_words = []

  vocab = vectorizer.get_feature_names_out()
  for k in range(numTop):
    top_idx = H[k].argsort()[::-1][:10]
    words = [vocab[i] for i in top_idx]
    topic_words.append(words)
  doc_topics = W.argmax(axis=1)
  return doc_topics, topic_words, docs
```

```python
keys = ["chatgpt", "gpt", "ai", "therapy", "therapist", "chat", "bot",
            "mental", "health", "friend", "friends", "emotional", "replying",
            "2024", "2023"]
negative_posts_file = pd.read_csv(negative_posts_path)
doc_topics, topic_words = run_nmf(negative_posts_file, keys, 5)
res = []
for i, words in enumerate(topic_words):
  res.append({"Topic": f"Topic {i}", "Words": ", ".join(words)})

nmf_neg_df = pd.DataFrame(res)
nmf_neg_df.to_csv("NMFNegativePostsTwitterResults.csv", index=False)
```
